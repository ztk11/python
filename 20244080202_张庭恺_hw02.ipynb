```python
#字符串的乐趣
def is_palindrome(s):
    s = "".join(ch.lower() for ch in s if ch.isalnum())
    return s == s[::-1]


def is_abecedarian(s):
    s = s.replace(" ", "").lower()
    return all(s[i] <= s[i + 1] for i in range(len(s) - 1))


def remove_vowels(s):
    return "".join(c for c in s if c.lower() not in "aeiou")
  
print("回文判断：")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))

print("\n字母顺序判断：")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))

print("\n删除元音：")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

```python
#列表的乐趣

def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("Input must be a list")
    return lst[::-1]


def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i] < seq[i - 1]:
            return False
    return True


print("列表反转：")
print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c"]))

print("\n升序判断：")
print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 3, 2]))
print(is_sorted([5, 5, 6, 7]))
```

```python
#作为向量的元组

def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("Vector must be a tuple")
    if not isinstance(s, (int, float)):
        raise TypeError("Scalar must be an integer or float")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("Vector elements must be numbers")

    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("Both vectors must be tuples")
    if len(t1) != len(t2):
        raise ValueError("Vectors must have the same length")
    if not all(isinstance(x, (int, float)) for x in t1 + t2):
        raise TypeError("Vector elements must be numbers")

    return float(sum(x * y for x, y in zip(t1, t2)))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if len(mx) == 0:
        return True

    n = len(mx[0])

    if not all(len(row) == n for row in mx):
        return False

    return all(
        isinstance(x, (int, float))
        for row in mx
        for x in row
    )


def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("Invalid matrix")

    if not isinstance(vec, tuple):
        raise TypeError("Vector must be a tuple")

    if not all(isinstance(x, (int, float)) for x in vec):
        raise TypeError("Vector elements must be numbers")

    if len(mx) > 0 and len(mx[0]) != len(vec):
        raise ValueError("Matrix columns must match vector length")

    return tuple(
        sum(x * y for x, y in zip(row, vec))
        for row in mx
    )


print("向量乘以标量：")
print(vec_scalar_mult((1, 2, 3), 2))

print("\n向量内积：")
print(vec_inner_product((1, 2, 3), (4, 5, 6)))

print("\n检查矩阵：")
print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3,))))

print("\n矩阵与向量相乘：")
print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
```

```python
#更多向量的乐趣
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("Dimension must be an integer")
        if dimension < 0:
            raise ValueError("Dimension must be non-negative")

        if elements is None:
            elements = (0,) * dimension

        if not isinstance(elements, (list, tuple)):
            raise TypeError("Elements must be a list or tuple")

        if len(elements) != dimension:
            raise ValueError("Dimension and number of elements must match")

        if not all(
            isinstance(x, (int, float)) and not isinstance(x, bool)
            for x in elements
        ):
            raise TypeError("Vector elements must be numbers")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _check_dimension(self, other):
        if not isinstance(other, Vector):
            raise TypeError("Operand must be a Vector")
        if self.dimension != other.dimension:
            raise ValueError("Vectors must have the same dimension")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_dimension(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_dimension(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_dimension(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_dimension(other)
        return self.elements >= other.elements

    def dot(self, other):
        self._check_dimension(other)
        return sum(
            x * y for x, y in zip(self.elements, other.elements)
        )

    def __mul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(
                self.dimension,
                tuple(x * other for x in self.elements)
            )

        if isinstance(other, Vector):
            self._check_dimension(other)
            return Vector(
                self.dimension,
                tuple(
                    x * y
                    for x, y in zip(self.elements, other.elements)
                )
            )

        raise TypeError("Operand must be a number or Vector")

    def __rmul__(self, other):
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
v3 = Vector(2, (2, 3))

print("v1 =", v1)
print("v2 =", v2)

print("\n比较运算：")
print("v1 == v3:", v1 == v3)
print("v1 < v2:", v1 < v2)
print("v1 <= v2:", v1 <= v2)
print("v1 > v2:", v1 > v2)
print("v1 >= v2:", v1 >= v2)

print("\n向量内积：")
print(v1.dot(v2))

print("\n标量乘法：")
print("2 * v1 =", 2 * v1)
print("v1 * 2 =", v1 * 2)

print("\n两个向量对应元素相乘：")
print("v1 * v2 =", v1 * v2)
```
