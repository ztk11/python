## 实验2：函数、条件、循环与递归


```python
def score_level(score):
    """90~100 优秀，80~<90 良好，60~<80 及格，0~<60 不及格。"""
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for value in (100, 95, 90, 85, 80, 60, 59, 0):
    print(value, "->", score_level(value))

for bad in (-1, 101):                   # 超出范围的输入
    try:
        score_level(bad)
    except ValueError as exc:
        print("ValueError:", exc)
```

```python
total_for = 0
for k in range(1, 101):
    total_for += k

total_while = 0
k = 1
while k <= 100:
    total_while += k
    k += 1

print("for   累加结果:", total_for)
print("while 累加结果:", total_while)
print("两者相等:", total_for == total_while, "| 公式 100*101//2 =", 100 * 101 // 2)
```

```python
def factorial_rec(n):
    """递归版阶乘，约定 0! = 1。"""
    if n <= 1:
        return 1
    return n * factorial_rec(n - 1)


def factorial_iter(n):
    """迭代版阶乘。"""
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


for n in (0, 1, 5, 10):
    r, it = factorial_rec(n), factorial_iter(n)
    print(f"n={n:2d}  递归={r}  迭代={it}  相等={r == it}")
```

```python
def summary(values):
    """返回元组（最小值, 最大值, 平均值）。"""
    return min(values), max(values), sum(values) / len(values)


low, high, mean = summary([85, 58, 92, 76, 60, 89])   # 元组解包
print("最小值 =", low, "| 最大值 =", high, "| 平均值 =", round(mean, 4))
```

```python
def sum_all(*args):
    total = 0
    for value in args:
        total += value
    return total


print("无参数  :", sum_all())
print("单个参数:", sum_all(7))
print("多个参数:", sum_all(1, 2, 3, 4, 5))
```

## 实验4：文件读写、异常处理与模块


```python
import os, sys, pickle

BASE_DIR = os.getcwd()                           # 查看当前工作目录
DATA_DIR = os.path.join(BASE_DIR, "exp01_data")  # 用 os.path.join 构造路径
os.makedirs(DATA_DIR, exist_ok=True)             # 目录已存在也不报错

DATA_FILE = os.path.join(DATA_DIR, "students.txt")
REPORT_FILE = os.path.join(DATA_DIR, "report.txt")
PICKLE_FILE = os.path.join(DATA_DIR, "students.pkl")

print("当前工作目录:", BASE_DIR)
print("数据目录    :", DATA_DIR)
print("目录是否存在:", os.path.isdir(DATA_DIR))
```

```python
STUDENT_TEXT = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(DATA_FILE, "w", encoding="utf-8") as fh:   # 写入文件
    fh.write(STUDENT_TEXT)

with open(DATA_FILE, "r", encoding="utf-8") as fh:   # 再读回文件
    content = fh.read()

print("字符数:", len(content), "| 行数:", len(content.splitlines()))
print(content)
```

```python
with open(DATA_FILE, "r", encoding="utf-8") as fh:
    lines = [line.strip() for line in fh if line.strip()]

print("表头:", lines[0])
for line in lines[1:4]:                 # 跳过表头，用字符串方法解析记录
    name, score = line.split(",")
    print(name, int(score))
```

```python
def parse_student(line):
    """把一行文本解析成 {"name": ..., "score": ...}，无效记录抛出 ValueError。"""
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数应为 2，实际为 {len(parts)}：{line.strip()!r}")
    name, raw_score = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError(f"姓名为空：{line.strip()!r}")
    try:
        score = int(raw_score)
    except ValueError:
        raise ValueError(f"成绩不是整数：{raw_score!r}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    return {"name": name, "score": score}


print(parse_student("Alice,85"))
```

```python
valid_records = []      # 有效记录
errors = []             # 无效记录：（行号, 原文, 原因）

with open(DATA_FILE, "r", encoding="utf-8") as fh:
    next(fh)                                  # 跳过表头
    for lineno, line in enumerate(fh, start=2):
        if not line.strip():
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:             # 记录错误后继续处理后续行
            errors.append((lineno, line.strip(), str(exc)))

print("有效记录数:", len(valid_records))
for rec in valid_records:
    print("   ", rec)
print("无效记录数:", len(errors))
for lineno, text, reason in errors:
    print(f"  第 {lineno} 行：{text!r} -> {reason}")
```

```python
n_valid, n_invalid = len(valid_records), len(errors)
mean = sum(rec["score"] for rec in valid_records) / n_valid
pass_rate = sum(1 for rec in valid_records if rec["score"] >= 60) / n_valid

with open(REPORT_FILE, "w", encoding="utf-8") as fh:   # "w"：清空原内容再写入
    fh.write(f"有效记录数: {n_valid}\n")
    fh.write(f"无效记录数: {n_invalid}\n")
    fh.write(f"平均分    : {mean:.2f}\n")
    fh.write(f"及格率    : {pass_rate:.1%}\n")

with open(REPORT_FILE, "r", encoding="utf-8") as fh:
    print("--- w 模式写入后 ---")
    print(fh.read())

with open(REPORT_FILE, "a", encoding="utf-8") as fh:   # "a"：在文件末尾追加
    fh.write("附注：w 模式会覆盖原有内容，a 模式只在末尾追加\n")

with open(REPORT_FILE, "r", encoding="utf-8") as fh:
    print("--- a 模式追加后 ---")
    print(fh.read())

with open(REPORT_FILE, "w", encoding="utf-8") as fh:   # 再用一次 w：原有内容全部丢失
    fh.write("现在文件里只剩这一行，可见 w 模式会清空原文件\n")

with open(REPORT_FILE, "r", encoding="utf-8") as fh:
    print("--- 再次用 w 模式写入后 ---")
    print(fh.read())
```

```python
with open(PICKLE_FILE, "wb") as fh:
    pickle.dump(valid_records, fh)          # 保存为二进制文件

with open(PICKLE_FILE, "rb") as fh:
    restored = pickle.load(fh)              # 读回

print("读回的数据与原数据相等:", restored == valid_records)
print(restored)
```

## 实验5：类、继承与对象复制


```python
class Student:
    """学生类：姓名 + 成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        state = "及格" if self.is_passed() else "不及格"
        return f"Student(姓名={self.name!r}, 成绩={self.score}, {state})"


s = Student("Alice", 85)
print(s)
print("是否及格:", s.is_passed())
```

```python
students = [Student(rec["name"], rec["score"]) for rec in valid_records]
for stu in students:
    print(f"{stu}  ->  是否及格={stu.is_passed()}")
```

```python
class GraduateStudent(Student):
    """研究生：在 Student 基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)        # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):                       # 重写父类方法
        return (f"GraduateStudent(姓名={self.name!r}, 成绩={self.score}, "
                f"研究方向={self.research_topic!r})")


g = GraduateStudent("Alice", 85, "数据挖掘")
print(g)
print("是否及格:", g.is_passed())
print("是 Student 的实例:", isinstance(g, Student))
```

```python
import copy


def make_original():
    return {"name": "Alice", "scores": [85, 90]}    # 每次新建原始对象


direct_obj = make_original()
direct_ref = direct_obj                 # 直接赋值：同一个对象
direct_ref["scores"].append(100)
print("直接赋值  :", direct_obj, "| 是同一对象:", direct_obj is direct_ref)

shallow_src = make_original()
shallow_obj = copy.copy(shallow_src)    # 浅复制：内层列表仍然共享
shallow_obj["scores"].append(100)
print("浅复制    : 原对象 =", shallow_src, "| 副本 =", shallow_obj,
      "| 内层列表共享:", shallow_src["scores"] is shallow_obj["scores"])

deep_src = make_original()
deep_obj = copy.deepcopy(deep_src)      # 深复制：完全独立
deep_obj["scores"].append(100)
print("深复制    : 原对象 =", deep_src, "| 副本 =", deep_obj,
      "| 内层列表共享:", deep_src["scores"] is deep_obj["scores"])
```
