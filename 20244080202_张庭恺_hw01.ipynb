## 欧几里得算法


```python
import math


def gcd(m, n):
    """用欧几里得算法求两个整数的最大公约数。"""
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m


for m, n in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({m}, {n}) = {gcd(m, n)}   （对照 math.gcd = {math.gcd(m, n)}）")
```

**问题3：如果参数为负数，函数会怎样？这种行为有意义吗？**



```python
def gcd_signed(m, n):      # 同一个循环，但不取绝对值
    while n != 0:
        m, n = n, m % n
    return m


for m, n in [(-24, 18), (24, -18), (-24, -18), (0, 5), (5, 0)]:
    print(f"gcd_signed({m:4d}, {n:4d}) = {gcd_signed(m, n):3d}   |   "
          f"gcd({m:4d}, {n:4d}) = {gcd(m, n):3d}")
```

## 近似欧拉数 e


```python
import math

print("math.exp(1) =", math.exp(1))


def euler_limit(n):
    """伯努利极限式：(1 + 1/n) ** n。"""
    return (1 + 1 / n) ** n


for n in (1, 2, 10, 100, 1000, 10 ** 6):
    print(f"n={n:>7}  euler_limit -> {euler_limit(n)!r}")
```

```python
def euler_infinite_sum(n):
    """公式（1）前 n 项之和：1 + 1/1! + 1/2! + 1/3! + ...，返回 float。

    euler_infinite_sum(0) -> 0.0
    euler_infinite_sum(1) -> 1.0
    euler_infinite_sum(4) -> 1 + 1 + 1/2 + 1/6
    """
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k          # k! = (k-1)! * k
        total += 1 / factorial
    return total


print("n=0 ->", euler_infinite_sum(0))
print("n=1 ->", euler_infinite_sum(1))
print("n=4 ->", euler_infinite_sum(4))
print("对照 1 + 1 + 1/2 + 1/6 =", 1 + 1 + 1 / 2 + 1 / 6)
print("n=10 ->", euler_infinite_sum(10), "| math.exp(1) =", math.exp(1))
```

```python
def euler_approx(epsilon):
    """用公式（1）的累加逼近 e，直到误差小于 epsilon。"""
    target = math.exp(1)
    n = 0
    approx = euler_infinite_sum(n)
    while abs(approx - target) > epsilon:
        n += 1
        approx = euler_infinite_sum(n)
    return approx


for eps in (1e-2, 1e-4, 1e-6, 1e-10):
    value = euler_approx(eps)
    print(f"epsilon={eps:<8g}  euler_approx={value!r}  误差={abs(value - math.exp(1)):.3e}")
```

```python
def print_euler_sum_table(n):
    """打印 k = 1..n 时的 euler_infinite_sum(k)，每行一个。"""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


print_euler_sum_table(5)
print("--- n = 12 ---")
print_euler_sum_table(12)
```

## 测试整数属性


```python
def is_power_of_2(n):
    """判断正整数 n 是否为 2 的幂（只使用 // 和 %）。

    n = 1 时是 2**0，也返回 True；n > 1 时反复除以 2，能除到 1 才是 2 的幂。
    """
    while n > 1:
        if n % 2 != 0:          # 大于 1 的奇数不可能是 2 的幂
            return False
        n = n // 2
    return n == 1


def is_power(b, n):
    """判断正整数 n 是否为 b 的幂（b、n 均为正整数）。"""
    if b == 1:                  # 1 的任何次幂都是 1
        return n == 1
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return n == 1
```

```python
print("is_power_of_2:")
for n in (1, 2, 3, 4, 6, 8, 16, 18, 1024, 1025):
    print(f"   {n:>5} -> {is_power_of_2(n)}")

print("is_power(b, n):")
for b, n in [(2, 8), (2, 12), (3, 9), (3, 27), (3, 28), (5, 1), (10, 1000),
             (10, 1001), (1, 1), (1, 2), (4, 64), (4, 16)]:
    print(f"   is_power({b}, {n}) -> {is_power(b, n)}")

# 与 is_power_of_2 交叉验证（1~199）
print("在 1~199 上与 is_power_of_2 一致:",
      all(is_power(2, n) == is_power_of_2(n) for n in range(1, 200)))
```
