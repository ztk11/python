```python
#第一部分：Python 正则表达式练习
def check_alphanumeric(s):
    return bool(re.fullmatch(r'[a-zA-Z0-9]+', s))


# 测试用例
tests = ["Hello123", "Python2026", "abc", "12345", "Hello@123", "你好123", ""]

for s in tests:
    print(s, ":", check_alphanumeric(s))
```

```python
def match_a_zero_or_more_b(s):
    return bool(re.fullmatch(r'ab*', s))


# 测试用例
tests = ["a", "ab", "abb", "abbb", "b", "ac", ""]

for s in tests:
    print(s, ":", match_a_zero_or_more_b(s))
```

```python
def match_a_one_or_more_b(s):
    return bool(re.fullmatch(r'ab+', s))


# 测试用例
tests = ["ab", "abb", "abbb", "a", "b", "abc"]

for s in tests:
    print(s, ":", match_a_one_or_more_b(s))
```

```python
def match_a_anything_b(s):
    return bool(re.fullmatch(r'a.*b', s))


# 测试用例
tests = ["ab", "acb", "a123b", "appleb", "ba", "abc"]

for s in tests:
    print(s, ":", match_a_anything_b(s))
```

```python
def remove_ip_leading_zeros(ip):
    parts = ip.split('.')
    return '.'.join(str(int(part)) for part in parts)


# 测试用例
tests = [
    "192.168.001.010",
    "010.000.001.255",
    "000.000.000.000",
    "192.168.1.1"
]

for ip in tests:
    print(ip, "->", remove_ip_leading_zeros(ip))
```

```python
def replace_space_underscore(s):
    return s.translate(str.maketrans(" _", "_ "))


# 测试用例
tests = [
    "hello world",
    "hello_world",
    "Python programming_language",
    "a_b c_d"
]

for s in tests:
    print(s, "->", replace_space_underscore(s))
```

```python
def convert_date(date):
    return re.sub(
        r'^(\d{4})-(\d{2})-(\d{2})$',
        r'\3-\2-\1',
        date
    )


# 测试用例
tests = [
    "2026-11-24",
    "2025-01-05",
    "2024-12-31"
]

for date in tests:
    print(date, "->", convert_date(date))
```

```python
def find_a_e_words(s):
    return re.findall(r'\b[aeAE]\w*\b', s)


# 测试用例
text = "apple orange elephant animal banana eagle computer"

print(find_a_e_words(text))
```

```python
def abbreviate_road(s):
    return re.sub(r'\bRoad\b', 'Rd', s)


# 测试用例
tests = [
    "123 Main Road",
    "Green Road",
    "Road to success",
    "Oxford Road"
]

for s in tests:
    print(s, "->", abbreviate_road(s))
```

```python
def find_long_words(s):
    return re.findall(r'\b[A-Za-z0-9]{4,}\b', s)


# 测试用例
text = "I love Python and data science 2026"

print(find_long_words(text))
```

```python
def remove_special_characters(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)


# 测试用例
tests = [
    "Hello, World! 123",
    "Python@2026",
    "A-B_C.D",
    "abc123"
]

for s in tests:
    print(s, "->", remove_special_characters(s))
```

```python
def remove_parentheses(s):
    return re.sub(r'\s*\([^)]*\)', '', s).strip()


# 测试用例
tests = [
    "example (.com)",
    "w3resource",
    "github (.com)",
    "stackoverflow (.com)"
]

for s in tests:
    print(s, "->", remove_parentheses(s))
```

```python
#第二部分：从网页检索气候数据
# 网页地址
URLS = {
    "Madison": "https://en.wikipedia.org/wiki/Madison,_Wisconsin",
    "Los Angeles": "https://en.wikipedia.org/wiki/Los_Angeles",
    "Boston": "https://en.wikipedia.org/wiki/Boston"
}

# 记录访问日期
print("访问日期：", datetime.now().strftime("%Y-%m-%d"))

# 选择 Madison 页面
url = URLS["Madison"]

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0 Safari/537.36"
}

response = requests.get(url, headers=headers, timeout=20)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

print("网页标题：", soup.title.get_text(strip=True))
print("网页读取成功！")
print("使用URL：", url)
```

```python
def normalize_text(text):
    return re.sub(r'\s+', ' ', text).strip()


def find_climate_table(soup):
    for table in soup.find_all("table"):
        text = normalize_text(table.get_text(" ", strip=True))

        if re.search(r'Climate data for', text, re.IGNORECASE):
            return table

        # 某些页面标题可能在表格附近
        previous = table.find_previous(
            ["caption", "th", "h2", "h3", "h4"]
        )

        if previous:
            title = normalize_text(previous.get_text(" ", strip=True))
            if re.search(r'Climate data for', title, re.IGNORECASE):
                return table

    return None


climate_table = find_climate_table(soup)

if climate_table:
    print("找到气候数据表格！")

    children = list(climate_table.children)

    print("直接子节点数量：", len(children))
    print("直接子节点标签：")

    for child in children:
        if getattr(child, "name", None):
            print(child.name)
else:
    print("没有找到气候数据表格")
```

```python
def retrieve_climate_table(s):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0 Safari/537.36"
    }

    response = requests.get(s, headers=headers, timeout=20)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    return find_climate_table(soup)
```

```python
for city, url in URLS.items():
    print("\n正在测试城市：", city)

    try:
        table = retrieve_climate_table(url)

        if table is not None:
            print("测试成功：找到气候数据表格")
            print("返回对象类型：", type(table))
        else:
            print("成功访问网页，但没有找到气候数据表格")

    except requests.RequestException as e:
        print("网页访问失败：", e)
```

```python
def list_climate_table_row_names(s):
    table = retrieve_climate_table(s)

    if table is None:
        return None

    row_names = []

    for tr in table.find_all("tr"):
        cells = tr.find_all(["th", "td"], recursive=False)

        if not cells:
            continue

        # 优先查找 scope="row" 的行标题
        row_header = tr.find(
            "th",
            attrs={"scope": "row"},
            recursive=False
        )

        # 部分网页没有 scope 属性
        if row_header is None:
            for cell in cells:
                if cell.name == "th":
                    text = normalize_text(cell.get_text(" ", strip=True))

                    if text and not re.search(
                        r'Climate data for|Jan|January|Feb|February|Mar|March|'
                        r'Apr|April|May|Jun|June|Jul|July|Aug|August|'
                        r'Sep|September|Oct|October|Nov|November|Dec|December',
                        text,
                        re.IGNORECASE
                    ):
                        row_header = cell
                        break

        if row_header is not None:
            name = normalize_text(row_header.get_text(" ", strip=True))

            if not name:
                continue

            # 排除标题和来源
            if re.search(
                r'Climate data for|Source|Sources|References',
                name,
                re.IGNORECASE
            ):
                continue

            if name not in row_names:
                row_names.append(name)

    return row_names
```

```python
url = URLS["Madison"]

names = list_climate_table_row_names(url)

print("城市：Madison")
print("气候数据行名：")

if names is None:
    print("没有找到气候数据表格")
else:
    for i, name in enumerate(names, 1):
        print(f"{i}. {name}")

    print("总行数：", len(names))
```
